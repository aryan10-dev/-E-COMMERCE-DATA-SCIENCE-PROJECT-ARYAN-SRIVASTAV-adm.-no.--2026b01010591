```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# E-COMMERCE DATA SCIENCE PROJECT
# DATA GENERATION, CLEANING AND ANALYSIS
# ==========================================

np.random.seed(42)

# STEP 1: CREATE DATASET

n = 500

data = {
    "Order_ID": range(1001, 1001 + n),

    "Customer_Name": np.random.choice(
        ["Rahul", "Aman", "Priya", "Sneha",
         "Rohit", "Anjali", "Karan", "Neha"],
        n
    ),

    "Product": np.random.choice(
        ["Laptop", "Mobile", "Shoes",
         "Watch", "Headphones", "T-Shirt"],
        n
    ),

    "Category": np.random.choice(
        ["Electronics", "Fashion", "Accessories"],
        n
    ),

    "Price": np.random.randint(500, 80000, n),

    "Quantity": np.random.randint(1, 6, n),

    "City": np.random.choice(
        ["Delhi", "Mumbai", "Lucknow",
         "Gorakhpur", "Pune", "Jaipur"],
        n
    ),

    "Payment_Method": np.random.choice(
        ["UPI", "Cash", "Credit Card", "Debit Card"],
        n
    ),

    "Order_Date": pd.date_range(
        start="2025-01-01",
        periods=n,
        freq="D"
    )
}

df = pd.DataFrame(data)

# STEP 2: ADD DIRTY DATA

df.loc[5, "Customer_Name"] = np.nan
df.loc[10, "City"] = np.nan
df.loc[15, "Price"] = np.nan
df.loc[20, "Quantity"] = -3
df.loc[25, "Price"] = -500
df.loc[30, "Payment_Method"] = np.nan

# Add duplicate records
df = pd.concat([df, df.iloc[:10]], ignore_index=True)

# Save raw dataset
df.to_csv("ecommerce_raw.csv", index=False)

print("RAW DATASET")
print(df.head(10))

print("\nDataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

# ==========================================
# STEP 3: DATA CLEANING
# ==========================================

clean_df = df.copy()

# Remove duplicates
clean_df.drop_duplicates(inplace=True)

# Fix negative and invalid values
clean_df.loc[clean_df["Price"] <= 0, "Price"] = np.nan

clean_df.loc[
    clean_df["Quantity"] <= 0, "Quantity"
] = np.nan

# Fill missing numerical values
clean_df["Price"] = clean_df["Price"].fillna(
    clean_df["Price"].median()
)

clean_df["Quantity"] = clean_df["Quantity"].fillna(
    clean_df["Quantity"].median()
)

# Fill missing categorical values
clean_df["Customer_Name"] = clean_df[
    "Customer_Name"
].fillna("Unknown")

clean_df["City"] = clean_df["City"].fillna(
    "Unknown"
)

clean_df["Payment_Method"] = clean_df[
    "Payment_Method"
].fillna("Unknown")

# Correct data types
clean_df["Price"] = clean_df["Price"].astype(float)

clean_df["Quantity"] = clean_df[
    "Quantity"
].astype(int)

clean_df["Order_Date"] = pd.to_datetime(
    clean_df["Order_Date"]
)

# Standardize text
for col in ["Customer_Name", "Product",
            "Category", "City", "Payment_Method"]:
    clean_df[col] = (
        clean_df[col].str.strip().str.title()
    )

# Assign correct category for each product
category_map = {
    "Laptop": "Electronics",
    "Mobile": "Electronics",
    "Headphones": "Electronics",
    "Shoes": "Fashion",
    "T-Shirt": "Fashion",
    "Watch": "Accessories"
}

clean_df["Category"] = clean_df[
    "Product"
].map(category_map)

# Calculate total sales
clean_df["Total_Sales"] = (
    clean_df["Price"] * clean_df["Quantity"]
)

# Extract year and month
clean_df["Year"] = clean_df[
    "Order_Date"
].dt.year

clean_df["Month"] = clean_df[
    "Order_Date"
].dt.month

# Save cleaned dataset
clean_df.to_csv(
    "ecommerce_cleaned.csv",
    index=False
)

print("\nCLEANED DATASET")
print(clean_df.head(10))

print("\nMissing Values After Cleaning:")
print(clean_df.isnull().sum())

print("\nDuplicate Records:",
      clean_df.duplicated().sum())

print("\nCleaned Dataset Shape:", clean_df.shape)

# ==========================================
# STEP 4: DATA ANALYSIS
# ==========================================

print("\n===== SALES ANALYSIS =====")

print("Total Orders:", len(clean_df))

print("Total Revenue: Rs.",
      round(clean_df["Total_Sales"].sum(), 2))

print("Average Order Value: Rs.",
      round(clean_df["Total_Sales"].mean(), 2))

print("\nSales By Product:")
print(
    clean_df.groupby("Product")[
        "Total_Sales"
    ].sum().sort_values(ascending=False)
)

print("\nSales By City:")
print(
    clean_df.groupby("City")[
        "Total_Sales"
    ].sum().sort_values(ascending=False)
)

print("\nPayment Method Usage:")
print(clean_df["Payment_Method"].value_counts())







print("\nPROJECT COMPLETED SUCCESSFULLY!")
print("Files Saved:")
print("1. ecommerce_raw.csv")
print("2. ecommerce_cleaned.csv")
```
